```python
import pandas as pd
import json

# Load the file
file_path = "top3_clicks_low_cpc_history.csv"
df = pd.read_csv(file_path)

# Step 1: Rank by highest clicks, then lowest CPC
ranked_df = df.sort_values(by=["clicks", "average_cpc_usd"], ascending=[False, True])

# Step 2: Keep only required columns + clicks + average_cpc_usd
required_columns = [
    "campaign_id",
    "ad_group_id",
    "Date_Day",
    "Negative_keywords_encoded",
    "Keyword_Quality_Estimated_Quality_Score",
    "resource_name",
    "ad_schedule_start_hour",
    "ad_schedule_end_hour",
    "Keyword_Quality_Keyword",
    "Keyword_performance_Keyword_Text",
    "keywords_meta_keyword_text",
    "search_term_search_term",
    "Keyword_Quality_KW_types",
    "bidding_strategy_bidding_strategy_type",
    "bid_Adjustments_device_type",
    "day_of_week",
    "geographic_report_geo_target_constant",
    "Device_performance_Device",
    "Keyword_performance_Keyword_Match_Type",
    "keywords_meta_match_type",
    "cost_usd",
    "clicks",
    "average_cpc_usd"
]

final_df = ranked_df[required_columns].head(5)

# Convert to JSON with columns as keys and lists of values
json_data = final_df.to_dict(orient="list")

# Show JSON in console
print(json.dumps(json_data, indent=2))

# Save JSON to file
with open("ranked_filtered_top5.json", "w") as f:
    json.dump(json_data, f, indent=2)
```
